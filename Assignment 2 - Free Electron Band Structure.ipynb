# Free Electron Band Structure 


### Question: 

Recall the derivation of H_k u_k = E_k u_k in class.

Consider switching the amplitude of the periodic potential slowly all the way to zero with the notion of periodicity "a" still intact.

Q1. Assume u_m = exp(i m 2pi x/a) in 1D, plot E_k for k = -6 pi/a to +6pi/a. Correlate m with band index.

Q2. Plot E_k for the same interval for periodicity "2a" and compare with your plot in Q1.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# I am taking ħ = 1,m=1, a = 1 for simplicity, so E(k) = ((k + m*2π/a)^2)/2
# k goes -6π/a to +6π/a, so we have values m=-3 to +3
# Plotting the bands labeled by m
a = 1.0
k = np.linspace(-6*np.pi/a, 6*np.pi/a, 5000)
h_cut = 1.055e-34
m = 9.11e-31

plt.figure(figsize=(12, 10))
for m in range(-3, 4):
    E = ((k + m * 2*np.pi/a)**2)/2
    plt.plot(k, E, label=f'm={m}')
plt.axvline(np.pi/a, linestyle='--', color='black', label='Ist Brillouin Zone')
plt.axvline(-np.pi/a, linestyle='--', color='black')
plt.xlabel("k")
plt.ylabel("E(k)")
plt.title("Free Electron Bands (Periodicity a)")
plt.grid()
plt.legend()
plt.show()
```

### Correlation of m and band index

In the Ist Brillouin zone (marked by black lines), we observe that the first energy band is given by m=0.

Second band is given by m=1 and m=-1

Third band is given by m=2 and m=-2
 
Fourth band is given by m=3 and m=-3

In general, the n-th band is given by m=n-1 and m=-(n-1)

```python
# Same as previous just that G=2π/a changes to G=π/a (considering periodicity of '2a'), so E(k) = (k + m*π/a)^2
a = 1.0
k = np.linspace(-6*np.pi/a, 6*np.pi/a, 5000)

plt.figure(figsize=(12, 10))
for m in range(-6, 7):
    E = ((k + m * np.pi/a)**2)/2
    plt.plot(k, E, label=f'm={m}')
plt.axvline(np.pi/2*a, linestyle='--', color='black', label='Ist Brillouin Zone')
plt.axvline(-np.pi/2*a, linestyle='--', color='black')
plt.xlabel("k")
plt.ylabel("E(k)")
plt.title("Free Electron Bands (Periodicity 2a)")
plt.grid()
plt.legend()
plt.show()
```

### Comparing the band structure for periodicity 'a' and '2a'

The band structure with periodicity 2a has twice the energy bands in the Ist Brillouin Zone, for the same range of k=-6pi/a to +6pi/a. 

The Brillouin zone is twice as small in this case.

To see the difference more clearly, we plot the two cases in the Ist Brillouin zone only.


```python
k = np.linspace(-np.pi/a, np.pi/a, 500)
h_cut = 1.055e-34
m = 9.11e-31

plt.figure(figsize=(12, 10))
for m in range(-3, 4):
    E = ((k + m * 2*np.pi/a)**2)/2
    plt.plot(k, E, label=f'm={m}')
plt.axvline(np.pi/a, linestyle='--', color='black', label='Ist Brillouin Zone')
plt.axvline(-np.pi/a, linestyle='--', color='black')
plt.xlabel("k")
plt.ylabel("E(k)")
plt.title("Free Electron Bands (Periodicity a)")
plt.grid()
plt.legend()
plt.show()
```

```python
k = np.linspace(-np.pi/2*a, np.pi/2*a, 5000)

plt.figure(figsize=(12, 10))
for m in range(-6, 7):
    E = ((k + m * np.pi/a)**2)/2
    plt.plot(k, E, label=f'm={m}')
plt.axvline(np.pi/2*a, linestyle='--', color='black', label='Ist Brillouin Zone')
plt.axvline(-np.pi/2*a, linestyle='--', color='black')
plt.xlabel("k")
plt.ylabel("E(k)")
plt.title("Free Electron Bands (Periodicity 2a)")
plt.grid()
plt.legend()
plt.show()
```

Therefore, we conclude that changing the periodicity from 'a' to '2a' shrinks the band structure and we observe twice as many bands in the zone shrinked by a factor of 2.
