# Free Electron Band Structure 


### Question: 

Recall the derivation of H_k u_k = E_k u_k in class.

Consider switching the amplitude of the periodic potential slowly all the way to zero with the notion of periodicity "a" still intact.

Q1. Assume u_m = exp(i m 2pi x/a) in 1D, plot E_k for k = -6 pi/a to +6pi/a. Correlate m with band index.

Q2. Plot E_k for the same interval for periodicity "2a" and compare with your plot in Q1.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# I am taking ħ = 1,m=1, a = 1 for simplicity, so E(k) = ((k + l*2π/a)^2)/2
# k goes -6π/a to +6π/a, so we have values m=-3 to +3
# Plotting the bands labeled by l
a = 1.0
k = np.linspace(-6*np.pi/a, 6*np.pi/a, 5000)
h_cut = 1.055e-34
m = 9.11e-31

plt.figure(figsize=(12, 10))
for l in range(-3, 4):
    E = ((k + l * 2*np.pi/a)**2)/2
    plt.plot(k, E, label=f'l={l}')
plt.axvline(np.pi/a, linestyle='--', color='black', label='Ist Brillouin Zone')
plt.axvline(-np.pi/a, linestyle='--', color='black')
plt.xlabel("k")
plt.ylabel("E(k)")
plt.title("Free Electron Bands (Periodicity a)")
plt.grid()
plt.legend()
plt.show()
```

```python
# Same as previous just that G=2π/a changes to G=π/a (considering periodicity of '2a'), so E(k) = (k + m*π/a)^2
a = 1.0
k = np.linspace(-6*np.pi/a, 6*np.pi/a, 5000)

plt.figure(figsize=(12, 10))
for m in range(-3, 4):
    E = ((k + m * np.pi/a)**2)/2
    plt.plot(k, E, label=f'm={m}')
plt.axvline(np.pi/2*a, linestyle='--', color='black', label='Ist Brillouin Zone')
plt.axvline(-np.pi/2*a, linestyle='--', color='black')
plt.xlabel("k")
plt.ylabel("E(k)")
plt.title("Free Electron Bands (Periodicity 2a)")
plt.grid()
plt.legend()
plt.show()
```
